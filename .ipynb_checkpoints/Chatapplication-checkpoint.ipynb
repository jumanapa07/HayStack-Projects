```python
from haystack import Pipeline,Document
from haystack.components.embedders import SentenceTransformersDocumentEmbedder
from haystack.document_stores.in_memory import InMemoryDocumentStore
from haystack.components.writers import DocumentWriter 
```

```python
documents = [
    Document(content="My name is Jean and I live in Paris."),
    Document(content="My name is Mark and I live in Berlin."),
    Document(content="My name is Giorgio and I live in Rome."),
    Document(content="My name is Marta and I live in Madrid."),
    Document(content="My name is Harry and I live in London."),
]
```

```python
embedder = SentenceTransformersDocumentEmbedder(model = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
document_store = InMemoryDocumentStore()
```

```python
documnet_writer = DocumentWriter(document_store=document_store)
```

```python
indexing_pipeline = Pipeline()

indexing_pipeline.add_component("embedder",embedder)
indexing_pipeline.add_component("documnet_writer",documnet_writer)


indexing_pipeline.connect("embedder.documents","documnet_writer")
```

```python
indexing_pipeline.run({"embedder":{"documents":documents}})
```

```python
from haystack.components.embedders import SentenceTransformersTextEmbedder
from haystack.components.retrievers.in_memory import InMemoryEmbeddingRetriever
from haystack.components.builders import ChatPromptBuilder
from haystack.components.generators.chat import OpenAIChatGenerator
from haystack.dataclasses import ChatMessage
```

```python
text_embedder = SentenceTransformersTextEmbedder(model = "sentence-transformers/all-MiniLM-L6-v2")
retreiver = InMemoryEmbeddingRetriever(document_store = document_store)
```

```python
template = [
    ChatMessage.from_user(
        """
        Answer the questions based on the given context.

        Context:
        {% for document in documents %}
            {{ document.content }}
        {% endfor %}
        
        Question: {{ question }}
        Answer:
        """
    )
]
builder = ChatPromptBuilder(template = template)
```

```python
from dotenv import load_dotenv
import os
load_dotenv(override=True)
api_key=os.getenv('OPENAI_API_KEY')
llm = OpenAIChatGenerator(model = "gpt-4o-mini")
```

```python
rag_pipeline = Pipeline()

rag_pipeline.add_component("text_embedder",text_embedder)
rag_pipeline.add_component("retreiver",retreiver)
rag_pipeline.add_component("builder",builder)
rag_pipeline.add_component("llm",llm)

rag_pipeline.connect("text_embedder","retreiver")
rag_pipeline.connect("retreiver.documents","builder.documents")
rag_pipeline.connect("builder.prompt","llm.messages")

```

```python
def rag_pipeline_func(query : str):
    result = rag_pipeline.run({"text_embedder":{"text":query},"builder":{"question":query}})
    return {"result" : result['llm']['replies'][0].content }
    
```

```python
WEATHER_INFO = {
    "Berlin": {"weather": "mostly sunny", "temperature": 7, "unit": "celsius"},
    "Paris": {"weather": "mostly cloudy", "temperature": 8, "unit": "celsius"},
    "Rome": {"weather": "sunny", "temperature": 14, "unit": "celsius"},
    "Madrid": {"weather": "sunny", "temperature": 10, "unit": "celsius"},
    "London": {"weather": "cloudy", "temperature": 9, "unit": "celsius"},
}


def get_current_weather(location: str):
    if location in WEATHER_INFO:
        return WEATHER_INFO[location]
    else:
        return {"weather": "sunny", "temperature": 21.8, "unit": "fahrenheit"}
```

```python

tools = [
    {
        "type": "function",
        "function": {
            "name": "rag_pipeline_func",
            "description": "Get information about where people live",
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {
                        "type": "string",
                        "description": "The query to use in the search. Infer this from the user's message. It should be a question or a statement",
                    }
                },
                "required": ["query"],
            },
        },
    },
    {
        "type": "function",
        "function": {
            "name": "get_current_weather",
            "description": "Get the current weather",
            "parameters": {
                "type": "object",
                "properties": {
                    "location": {"type": "string", "description": "The city and state, e.g. San Francisco, CA"}
                },
                "required": ["location"],
            },
        },
    },
]
```

```python
# from haystack.components.generators.utils import print_streaming_chunk
# messages = [
#     ChatMessage.from_system("Don't make assumptions about what values to plug into functions. Ask for clarification if a user request is ambiguous."),
#     ChatMessage.from_user("Can you tell me where Mark lives?")
# ]

# client = OpenAIChatGenerator(model = "gpt-4o-mini" ,streaming_callback = print_streaming_chunk )
# response = client.run(messages = messages , generation_kwargs = {"tools":tools})
# print(response)

from haystack.dataclasses import ChatMessage
from haystack.components.generators.chat import OpenAIChatGenerator
from haystack.components.generators.utils import print_streaming_chunk

messages = [
    ChatMessage.from_system(
        "Don't make assumptions about what values to plug into functions. Ask for clarification if a user request is ambiguous."
    ),
    ChatMessage.from_user("Can you tell me where Mark lives?"),
]

chat_generator = OpenAIChatGenerator(model="gpt-4o-mini", streaming_callback=print_streaming_chunk)
response = chat_generator.run(messages=messages, generation_kwargs={"tools": tools})
```

```python
# import json 
# tool_call = response['replies'][0].tool_call
# function_name=(response['replies'][0].tool_call.tool_name)
# function_args = (response['replies'][0].tool_call.arguments)
# available_functions = {"rag_pipeline_func": rag_pipeline_func, "get_current_weather": get_current_weather}
# function_to_call = available_functions[function_name]
# function_response = function_to_call(**function_args)
# print("Function Response:", function_response)
import json

## Parse function calling information
function_call = json.loads(response["replies"][0].text)[0]
function_name = function_call["function"]["name"]
function_args = json.loads(function_call["function"]["arguments"])
print("Function Name:", function_name)
print("Function Arguments:", function_args)

## Find the correspoding function and call it with the given arguments
available_functions = {"rag_pipeline_func": rag_pipeline_func, "get_current_weather": get_current_weather}
function_to_call = available_functions[function_name]
function_response = function_to_call(**function_args)
print("Function Response:", function_response)
```

```python
# messages.append(ChatMessage.from_function(content=json.dumps(function_response), name=function_name))
# function_message = ChatMessage.from_tool( tool_result = str(function_response),origin=tool_call)
# print(function_message)
# messages.append(function_message)

from haystack.dataclasses import ChatMessage

function_message = ChatMessage.from_function(content=json.dumps(function_response), name=function_name)
messages.append(function_message)

response = chat_generator.run(messages=messages, generation_kwargs={"tools": tools})
```
